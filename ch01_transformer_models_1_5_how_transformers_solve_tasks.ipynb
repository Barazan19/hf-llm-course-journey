## 1. Transformer Models for Language

### How Language Models Work
Secara umum, Language Models (Model Bahasa) dilatih untuk memprediksi kata berikutnya dalam sebuah kalimat berdasarkan konteks kata-kata sebelumnya (atau di sekitarnya). Proses ini melatih model untuk memahami tata bahasa, struktur, dan makna dari bahasa manusia.

### Types of Language Models
Model-model Transformer biasanya dibagi menjadi tiga kategori utama berdasarkan arsitekturnya:
1. **Encoder-only (Auto-encoding models):** Cocok untuk tugas yang membutuhkan pemahaman mendalam terhadap seluruh kalimat (misalnya BERT). Sangat bagus untuk klasifikasi teks atau ekstraksi informasi.
2. **Decoder-only (Auto-regressive models):** Cocok untuk menghasilkan teks baru (misalnya GPT). Model ini memprediksi kata berikutnya satu per satu secara berurutan.
3. **Encoder-Decoder (Sequence-to-sequence models):** Menggabungkan keduanya (misalnya T5, BART). Sangat cocok untuk tugas translasi atau ringkasan (summarization) di mana input teks diubah menjadi representasi tertentu lalu diterjemahkan menjadi teks baru.

### A. Text Generation (Decoder-only)
Tugas ini bertujuan untuk memproduksi teks kelanjutan dari sebuah prompt input.

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
prompt = "Today, I started learning about Hugging Face Transformers and"
result = generator(prompt, max_length=50, num_return_sequences=1)

print(result[0]['generated_text'])
```

### B. Text Classification (Encoder-only)
Menentukan kategori atau sentimen dari sebuah kalimat.

```python
classifier = pipeline("sentiment-analysis")
res = classifier("I love learning NLP with Hugging Face, it is so intuitive!")
print(res)
```

### C. Token Classification (NER)
Mengklasifikasikan setiap token/kata individual, misalnya mendeteksi nama orang, lokasi, atau organisasi (Named Entity Recognition).

```python
ner = pipeline("ner", aggregation_strategy="simple")
res = ner("My name is Sarah and I work at Hugging Face in New York.")
for entity in res:
    print(f"Entity: {entity['word']} -> Label: {entity['entity_group']} (Score: {entity['score']:.2f})")
```

### D. Question Answering
Mengekstrak jawaban dari sebuah teks dokumen (konteks) berdasarkan pertanyaan yang diajukan.

```python
import torch
from transformers import AutoTokenizer, AutoModelForQuestionAnswering

# Menggunakan manual tokenizer dan model untuk menghindari isu task registration di pipeline
model_name = "distilbert-base-cased-distilled-squad"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForQuestionAnswering.from_pretrained(model_name)

context = "Transformers is an open-source library created by Hugging Face in 2018."
question = "Who created Transformers?"

# 1. Tokenisasi input teks
inputs = tokenizer(question, context, return_tensors="pt")

# 2. Ambil output logits dari model
with torch.no_grad():
    outputs = model(**inputs)

# 3. Cari indeks token untuk awal dan akhir jawaban
answer_start_index = outputs.start_logits.argmax()
answer_end_index = outputs.end_logits.argmax() + 1

# 4. Decode hasil token menjadi kata
predict_answer_tokens = inputs.input_ids[0, answer_start_index : answer_end_index]
decoded_answer = tokenizer.decode(predict_answer_tokens)

print(f"Question: {question}")
print(f"Answer: {decoded_answer}")
```

### E. Summarization
Merangkum teks yang panjang menjadi lebih ringkas dan padat.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Memakai model & tokenizer secara langsung agar tidak terkendala registrasi pipeline
model_name = "facebook/bart-large-cnn"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = """
An oil spill is the release of a liquid petroleum hydrocarbon into the environment, especially the marine ecosystem, due to human activity, and is a form of pollution. The term is mostly applied to marine oil spills, where oil is released into the ocean or coastal waters, but spills may also occur on land. Oil spills may be due to releases of crude oil from tankers, offshore platforms, drilling rigs and wells, as well as spills of refined petroleum products (such as gasoline, diesel) and their by-products.
"""

# Tokenisasi input
inputs = tokenizer(text, max_length=1024, truncation=True, return_tensors="pt")

# Generate ringkasan
summary_ids = model.generate(inputs["input_ids"], num_beams=4, min_length=20, max_length=50, early_stopping=True)
summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

print("Summary:")
print(summary)
```

### F. Translation
Menerjemahkan teks dari satu bahasa ke bahasa lain.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Menggunakan model penerjemahan khusus MarianMT secara manual
model_name = "Helsinki-NLP/opus-mt-en-id"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "Learning machine learning is exciting and very rewarding."

# Tokenisasi input
inputs = tokenizer(text, return_tensors="pt")

# Generate hasil terjemahan
translated_ids = model.generate(**inputs)
translation = tokenizer.decode(translated_ids[0], skip_special_tokens=True)

print(f"Original: {text}")
print(f"Translation: {translation}")
```
